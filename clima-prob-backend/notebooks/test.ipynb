```python
import io, re
import pandas as pd
import numpy as np
import requests

def parse_giovanni_csv(csv_text: str) -> pd.DataFrame:
    """
    Encuentra dinámicamente la línea de encabezado que contiene 'Timestamp'
    y parsea la tabla completa a partir de ahí. Devuelve un DataFrame con
    índice de tiempo (UTC) y una única columna de valores (nombre real).
    """
    # 1) Localiza la fila del header
    lines = csv_text.splitlines()
    header_idx = None
    for i, line in enumerate(lines):
        if re.match(r'^\s*Timestamp\s*,', line):
            header_idx = i
            break
    if header_idx is None:
        raise RuntimeError("No se encontró la línea de encabezado 'Timestamp' en el CSV de Giovanni.")

    # 2) Corta el texto desde el header y carga con pandas
    data_text = "\n".join(lines[header_idx:])
    df = pd.read_csv(io.StringIO(data_text))

    # 3) Normaliza nombres
    if "Timestamp" not in df.columns:
        # Algunos despliegues pueden devolver 'Time' o similar
        time_col = [c for c in df.columns if c.lower().startswith("time")][0]
        df = df.rename(columns={time_col: "Timestamp"})
    val_cols = [c for c in df.columns if c != "Timestamp"]
    if not val_cols:
        raise RuntimeError("No se encontró columna de valores en el CSV de Giovanni.")
    val_col = val_cols[0]

    # 4) Índice temporal UTC
    df["Timestamp"] = pd.to_datetime(df["Timestamp"], utc=True, errors="coerce")
    df = df.dropna(subset=["Timestamp"]).set_index("Timestamp").sort_index()

    # 5) Asegura tipo numérico
    df[val_col] = pd.to_numeric(df[val_col], errors="coerce")
    return df[[val_col]]
```

```python
# === Giovanni Time Series: token + fetch genérico + parsers y agregadores ===
import netrc, os, io, time, math
import requests
from requests.auth import HTTPBasicAuth
import pandas as pd
import numpy as np

SIGNIN_URL = "https://api.giovanni.earthdata.nasa.gov/signin"
TS_URL     = "https://api.giovanni.earthdata.nasa.gov/timeseries"

def _giovanni_auth():
    """Devuelve (user, pass) desde ~/.netrc o variables de entorno."""
    user = os.getenv("EARTHDATA_USERNAME")
    pwd  = os.getenv("EARTHDATA_PASSWORD")
    if user and pwd:
        return user, pwd
    try:
        host = 'urs.earthdata.nasa.gov'
        nrc = netrc.netrc()
        login, _, password = nrc.hosts[host]
        return login, password
    except Exception as e:
        raise RuntimeError(
            "No encontré credenciales. Crea ~/.netrc o exporta "
            "EARTHDATA_USERNAME/EARTHDATA_PASSWORD."
        ) from e

def giovanni_token():
    """Obtiene el token de sesión de Giovanni (texto plano)."""
    user, pwd = _giovanni_auth()
    r = requests.get(SIGNIN_URL, auth=HTTPBasicAuth(user, pwd), allow_redirects=True, timeout=30)
    r.raise_for_status()
    return r.text.replace('"','').strip()

def giovanni_timeseries(data_id: str, lat: float, lon: float, start_iso: str, end_iso: str, token: str) -> pd.DataFrame:
    """
    Llama a /timeseries con un 'data' (p. ej. 'GLDAS_NOAH025_3H_2_1_Tair_f_inst')
    y devuelve DataFrame con columnas ['Timestamp', <param_name>] en UTC.
    """
    params = {
        "data": data_id,
        "location": f"[{lat},{lon}]",
        "time": f"{start_iso}/{end_iso}",
    }
    headers = {"authorizationtoken": token}
    r = requests.get(TS_URL, params=params, headers=headers, timeout=120)
    r.raise_for_status()
    csv_text = r.text
    print(csv_text)

    # Parse del CSV: los primeros ~13 renglones son metadatos clave, luego 'Timestamp,<param>'
    with io.StringIO(csv_text) as f:
        headers_kv = {}
        header_lines = []
        # leemos hasta encontrar la línea de encabezado real ("Timestamp,")
        for _ in range(40):  # margen amplio
            pos = f.tell()
            line = f.readline()
            if not line:
                break
            if line.startswith("Timestamp (UTC),"):
                # retrocede al inicio de esta línea para que pandas lea desde acá
                f.seek(pos)
                break
            try:
                key, value = line.split(",", 1)
                headers_kv[key.strip()] = value.strip()
                header_lines.append(line)
            except ValueError:
                header_lines.append(line)

        # Nombre de columna (param_name) si viene en header
        colname = headers_kv.get("param_name", "value")
        df = pd.read_csv(f, header=0, names=("Timestamp", colname))
        df["Timestamp"] = pd.to_datetime(df["Timestamp"], utc=True, errors="coerce")
        df = df.dropna(subset=["Timestamp"]).set_index("Timestamp").sort_index()

    # añade metadatos útiles
    df.attrs["headers"] = headers_kv
    return df

# --- Conversión y agregación diaria ---
def K_to_C(x): return x - 273.15

def daily_agg(series: pd.Series, how: str) -> pd.Series:
    if how == "max":  return series.resample("1D").max()
    if how == "min":  return series.resample("1D").min()
    if how == "mean": return series.resample("1D").mean()
    if how == "sum":  return series.resample("1D").sum()
    raise ValueError("how debe ser 'max|min|mean|sum'")

# Saturation vapor pressure (Pa) Magnus-Tetens sobre agua, T en °C
def es_pa_from_Tc(Tc: pd.Series) -> pd.Series:
    es_hPa = 6.112 * np.exp(17.67 * Tc / (Tc + 243.5))
    return es_hPa * 100.0

# RH% a partir de q (kg/kg), P (Pa), T (K)
def rh_from_q_p_t(q: pd.Series, P: pd.Series, T_K: pd.Series) -> pd.Series:
    # q = 0.622 e / (P - 0.378 e)  ->  e = q P / (0.622 + 0.378 q)
    e = (q * P) / (0.622 + 0.378 * q)  # Pa
    Tc = K_to_C(T_K)
    es = es_pa_from_Tc(Tc)
    rh = (e / es) * 100.0
    return rh.clip(lower=0, upper=100)

# Heat Index (Rothfusz/NWS) – entradas T en °C y RH% ; salida °C
def heat_index_C(Tc: pd.Series, RH: pd.Series) -> pd.Series:
    # convierto a F para fórmula NWS
    Tf = Tc * 9.0/5.0 + 32.0
    # coeficientes Rothfusz
    c = {
        "c1": -42.379, "c2": 2.04901523, "c3": 10.14333127, "c4": -0.22475541,
        "c5": -6.83783e-3, "c6": -5.481717e-2, "c7": 1.22874e-3, "c8": 8.5282e-4, "c9": -1.99e-6
    }
    HI_f = (c["c1"] + c["c2"]*Tf + c["c3"]*RH + c["c4"]*Tf*RH +
            c["c5"]*Tf*Tf + c["c6"]*RH*RH + c["c7"]*Tf*Tf*RH +
            c["c8"]*Tf*RH*RH + c["c9"]*Tf*Tf*RH*RH)
    # Ajustes opcionales a bajos rangos podrían aplicarse, pero para MVP basta
    HI_c = (HI_f - 32.0) * 5.0/9.0
    return HI_c
```

```python
def gldas_daily_series(lat, lon, start_iso, end_iso):
    token = giovanni_token()

    # 3-hourly series
    t_df = giovanni_timeseries("GLDAS_NOAH025_3H_2_1_Tair_f_inst",  lat, lon, start_iso, end_iso, token)
    w_df = giovanni_timeseries("GLDAS_NOAH025_3H_2_1_Wind_f_inst",  lat, lon, start_iso, end_iso, token)
    q_df = giovanni_timeseries("GLDAS_NOAH025_3H_2_1_Qair_f_inst",  lat, lon, start_iso, end_iso, token)
    p_df = giovanni_timeseries("GLDAS_NOAH025_3H_2_1_Psurf_f_inst", lat, lon, start_iso, end_iso, token)

    # Nombres reales devueltos (por si Giovanni usa encabezados distintos)
    tcol = list(t_df.columns)[0]
    wcol = list(w_df.columns)[0]
    qcol = list(q_df.columns)[0]
    pcol = list(p_df.columns)[0]

    # Conversión y agregación a diario
    T_K   = t_df[tcol]
    T_C   = K_to_C(T_K)
    WS    = w_df[wcol]
    Qair  = q_df[qcol]
    Psurf = p_df[pcol]

    # RH% horario y HI horario
    RH_pct  = rh_from_q_p_t(Qair, Psurf, T_K)
    HI_C_hr = heat_index_C(T_C, RH_pct)

    out = pd.DataFrame({
        "Tmax_C": daily_agg(T_C, "max"),
        "Tmin_C": daily_agg(T_C, "min"),
        "WS_ms":  daily_agg(WS,   "mean"),
        "RH_pct": daily_agg(RH_pct, "mean"),
        "HI_C":   daily_agg(HI_C_hr, "max"),
    })
    print(out)
    return out
```

```python
IMERG_DAILY_CANDIDATES = [
    # V07 explícito (recomendado):
    "GPM_3IMERGDF_07_precipitation",
    # Retrocompatibilidad o despliegues intermedios:
    "GPM_3IMERGDF_07_precipitationCal",
    # Sin versión explícita (algunos entornos lo aceptan):
    "GPM_3IMERGDF_precipitation",
    "GPM_3IMERGDF_precipitationCal",
]

def imerg_daily_series(lat, lon, start_iso, end_iso, *, verbose=True) -> pd.Series:
    """
    Devuelve serie diaria de precipitación (mm/day) usando Giovanni Time Series.
    Prueba varias cadenas 'data' para tolerar diferencias entre despliegues.
    """
    token = giovanni_token()
    last_err = None
    tried = []
    for data_id in IMERG_DAILY_CANDIDATES:
        try:
            if verbose:
                print(f"[IMERG] Probando data='{data_id}'")
            df = giovanni_timeseries(data_id, lat, lon, start_iso, end_iso, token)

            col = df.columns[0]
            s = df[col].rename("P_mmday").astype(float)

            # Asegura frecuencia diaria (Giovanni suele devolver timestamps diarios ya)
            s = s.resample("1D").mean()

            full_index = pd.date_range(start=start_iso[:10], end=end_iso[:10], freq="D", tz="UTC")
            s = s.reindex(full_index)  # alinea a todos los días solicitados

            if verbose:
                heads = df.attrs.get("headers", {})
                print(f"[IMERG] OK con '{data_id}' | unidad={heads.get('unit','?')} | param={heads.get('param_name', col)}")

            return s
        except requests.HTTPError as e:
            last_err = e
            tried.append(data_id)
        except Exception as e:
            last_err = e
            tried.append(data_id)

    raise RuntimeError(
        "No se pudo obtener IMERG Daily desde Giovanni.\n"
        f"Intentos fallidos: {tried}\n"
        f"Último error: {repr(last_err)}\n"
        "Sugerencia: verifica en la interfaz web de Giovanni el nombre exacto del parámetro para 3IMERGDF V07."
    )

```

```python
import numpy as np
import pandas as pd
from dataclasses import dataclass
from typing import Dict, Tuple

_DOY_CUM = np.array([0, 31, 59, 90, 120, 151, 181, 212, 243, 273, 304, 334], dtype=int)

# ---- Umbrales por defecto (ajústalos) ----
@dataclass
class Thresholds:
    very_hot_Tmax_C: float = 32.0
    very_cold_Tmin_C: float = 0.0
    very_windy_speed_ms: float = 10.0
    very_wet_precip_mmday: float = 20.0
    very_uncomfortable_HI_C: float = 32.0  # opcional si no tienes HI_C


def _ensure_daily_index(df: pd.DataFrame) -> pd.DataFrame:
    """Ordena, quita duplicados por fecha y asegura índice diario (UTC o naivo)"""
    d = df.copy().sort_index()
    d = d[~d.index.duplicated(keep="first")]
    # si el índice es tz-aware, lo dejamos así; si no, ok
    return d

def _doy365(idx: pd.DatetimeIndex) -> np.ndarray:
    """
    DOY en calendario de 365 días. 29-feb se clampa a 28-feb.
    Funciona con índices tz-aware o naivos (siempre usa mes/día del índice).
    """
    months = idx.month.values
    days   = idx.day.values
    # 29-feb -> 28-feb
    days = np.where((months == 2) & (days == 29), 28, days)
    return _DOY_CUM[months - 1] + days

def _window_mask(idx: pd.DatetimeIndex, date_of_interest: pd.Timestamp, window_days: int) -> np.ndarray:
    """
    Ventana ±K días alrededor del DOY objetivo (circular en 365).
    Usa UTC si el índice es tz-aware para mantener consistencia.
    """
    # Asegura UTC
    idx_utc = idx.tz_convert("UTC") if getattr(idx, "tz", None) is not None else idx
    doi = pd.to_datetime(date_of_interest)
    doi = doi.tz_convert("UTC") if doi.tzinfo is not None else doi.tz_localize("UTC")

    doy_series = _doy365(idx_utc)
    doy0 = _doy365(pd.DatetimeIndex([doi]))[0]

    dist = np.abs(doy_series - doy0)
    dist = np.minimum(dist, 365 - dist)  # circular
    return dist <= window_days

def _wilson_interval(k: int, n: int, z: float = 1.96) -> Tuple[float, float, float]:
    """Intervalo de Wilson 95%: retorna (p, lo, hi)."""
    if n == 0:
        return (np.nan, np.nan, np.nan)
    p = k / n
    denom = 1 + z**2 / n
    center = p + z**2/(2*n)
    adj = z*np.sqrt((p*(1-p) + z**2/(4*n)) / n)
    lo = (center - adj) / denom
    hi = (center + adj) / denom
    return (p, max(0.0, lo), min(1.0, hi))
```

```python
def _labels_from_thresholds(df: pd.DataFrame, thr: Thresholds) -> Dict[str, pd.Series]:
    lab = {}
    if "Tmax_C" in df:
        lab["very_hot"]  = (df["Tmax_C"] >= thr.very_hot_Tmax_C)
    if "Tmin_C" in df:
        lab["very_cold"] = (df["Tmin_C"] <= thr.very_cold_Tmin_C)
    if "WS_ms" in df:
        lab["very_windy"] = (df["WS_ms"] >= thr.very_windy_speed_ms)
    if "P_mmday" in df:
        lab["very_wet"]  = (df["P_mmday"] >= thr.very_wet_precip_mmday)
    if "HI_C" in df:
        lab["very_uncomfortable"] = (df["HI_C"] >= thr.very_uncomfortable_HI_C)
    return lab
```

```python
def empirical_probabilities(
    df_daily: pd.DataFrame,
    date_of_interest: str,
    thresholds: Thresholds,
    window_days: int = 7
) -> Dict[str, Dict[str, float]]:
    """
    Devuelve dict por categoría: {'prob': p, 'lo': lo, 'hi': hi, 'n': n, 'k': k}
    usando frecuencia de excedencia y Wilson 95% en ventana ±K días (todos los años).
    """
    df = _ensure_daily_index(df_daily)
    doi = pd.to_datetime(date_of_interest)
    mask = _window_mask(df.index, doi, window_days)
    dfw = df.loc[mask].dropna(how="all")
    labs = _labels_from_thresholds(dfw, thresholds)

    out = {}
    for name, y in labs.items():
        yy = y.dropna()
        n = int(yy.shape[0])
        k = int(yy.sum())
        p, lo, hi = _wilson_interval(k, n, z=1.96)
        out[name] = {"prob": float(p if np.isfinite(p) else 0.0),
                     "lo": float(lo if np.isfinite(lo) else 0.0),
                     "hi": float(hi if np.isfinite(hi) else 0.0),
                     "n": n, "k": k}
    return out
```

```python
from sklearn.linear_model import LogisticRegression

def _years_float(idx: pd.DatetimeIndex) -> np.ndarray:
    """Asegura UTC si hay tz y devuelve años como float ndarray."""
    if getattr(idx, "tz", None) is not None:
        idx = idx.tz_convert("UTC")
    return idx.year.values.astype(float)

def logistic_probabilities(
    df_daily: pd.DataFrame,
    date_of_interest: str,
    thresholds: Thresholds,
    window_days: int = 7,
    min_pos: int = 5,
    min_neg: int = 5,
) -> Dict[str, Dict[str, float]]:
    """
    logit(y) ~ 1 + year_norm  (ventana ±K días alrededor del DOY en todos los años).
    Si hay pocos positivos/negativos -> fallback empírico (Wilson).
    """
    df = _ensure_daily_index(df_daily)
    doi = pd.to_datetime(date_of_interest)
    mask = _window_mask(df.index, doi, window_days)
    sub = df.loc[mask].dropna(how="all")
    labs = _labels_from_thresholds(sub, thresholds)

    # años de la submuestra (ndarray, no Index)
    years_all = _years_float(sub.index)
    if years_all.size == 0:
        return {name: {"prob": np.nan, "n": 0, "k": 0, "engine": "empty"} for name in labs}

    year0 = float(np.mean(years_all))
    target_year = float(doi.year)
    X_pred = np.array([[target_year - year0]])

    out = {}
    for name, y in labs.items():
        yy = y.dropna().astype(int)
        if yy.empty:
            out[name] = {"prob": np.nan, "n": 0, "k": 0, "engine": "empty"}
            continue

        # Alinear X e y por índice
        common_idx = yy.index.intersection(sub.index)
        yi = yy.loc[common_idx].values
        Xi = (_years_float(common_idx) - year0).reshape(-1, 1)

        pos = int(yi.sum()); neg = int(yi.shape[0] - pos)
        n = pos + neg
        if pos < min_pos or neg < min_neg:
            p, lo, hi = _wilson_interval(pos, n)
            out[name] = {
                "prob": float(0.0 if not np.isfinite(p) else p),
                "n": n, "k": pos, "engine": "empirical_fallback"
            }
            continue

        clf = LogisticRegression(class_weight="balanced", solver="lbfgs")
        clf.fit(Xi, yi)
        p = float(clf.predict_proba(X_pred)[0, 1])
        out[name] = {"prob": p, "n": n, "k": pos, "engine": "logistic_year_trend"}

    return out
```

```python
def compute_probabilities(
    df_daily: pd.DataFrame,
    date_of_interest: str,
    thresholds: Dict[str, float],
    window_days: int = 7,
    engine: str = "empirical"  # "empirical" | "logistic"
) -> Dict[str, float]:
    thr = Thresholds(**thresholds)
    if engine == "logistic":
        res = logistic_probabilities(df_daily, date_of_interest, thr, window_days=window_days)
    else:
        res = empirical_probabilities(df_daily, date_of_interest, thr, window_days=window_days)
    # sólo el número (0–1) para cada categoría disponible
    return {k: float(v.get("prob", np.nan)) for k, v in res.items()}
```

```python
import numpy as np
import pandas as pd

# --- 1) Umbrales adaptativos a partir del dataset (ventana ±K días) ---
def make_thresholds_from_df(
    df: pd.DataFrame,
    date_of_interest: str,
    window_days: int = 7,
    *,
    p_hot=0.90, p_cold=0.10, p_windy=0.90, p_wet=0.90, p_hi=0.90,
    wet_floor_mm=1.0, wind_floor_ms=2.0,
    default_fallbacks=None
) -> dict:
    """
    Calcula umbrales por percentil en la ventana ±K días alrededor del DOY.
    Retorna dict con claves compatibles con compute_probabilities(...).
    """
    if default_fallbacks is None:
        default_fallbacks = {
            "very_hot_Tmax_C": 30.0,     # fallback más realista que 32 para climas templados
            "very_cold_Tmin_C": 5.0,
            "very_windy_speed_ms": 6.0,  # media diaria
            "very_wet_precip_mmday": 10.0,
            "very_uncomfortable_HI_C": 32.0,
        }

    # Ventana ±K días (usa tus helpers arreglados _window_mask/_doy365)
    doi = pd.to_datetime(date_of_interest)
    mask = _window_mask(df.index, doi, window_days)
    sub = df.loc[mask]

    def pct(series, p):
        s = series.dropna()
        return float(np.nanpercentile(s, p*100)) if s.size else np.nan

    thr = {}

    if "Tmax_C" in sub:
        v = pct(sub["Tmax_C"], p_hot)
        thr["very_hot_Tmax_C"] = float(v) if np.isfinite(v) else default_fallbacks["very_hot_Tmax_C"]

    if "Tmin_C" in sub:
        v = pct(sub["Tmin_C"], p_cold)
        thr["very_cold_Tmin_C"] = float(v) if np.isfinite(v) else default_fallbacks["very_cold_Tmin_C"]

    if "WS_ms" in sub:
        v = pct(sub["WS_ms"], p_windy)
        # piso mínimo para que no quede ridículo
        if np.isfinite(v):
            v = max(v, wind_floor_ms)
        thr["very_windy_speed_ms"] = float(v) if np.isfinite(v) else default_fallbacks["very_windy_speed_ms"]

    if "P_mmday" in sub:
        v = pct(sub["P_mmday"], p_wet)
        if np.isfinite(v):
            v = max(v, wet_floor_mm)  # al menos 1 mm/d aseguro “día lluvioso”
        thr["very_wet_precip_mmday"] = float(v) if np.isfinite(v) else default_fallbacks["very_wet_precip_mmday"]

    if "HI_C" in sub:
        v = pct(sub["HI_C"], p_hi)
        thr["very_uncomfortable_HI_C"] = float(v) if np.isfinite(v) else default_fallbacks["very_uncomfortable_HI_C"]

    return thr


# --- 2) Umbrales desde el dict de stats que ya tienes (p10/p50/p90) ---
def make_thresholds_from_stats(stats: dict, *, wet_floor_mm=1.0, wind_floor_ms=2.0,
                               defaults=None) -> dict:
    """
    stats: dict como el que mostraste (con 'p10','p50','p90')
    Usa p90 para 'hot/windy/wet/HI' y p10 para 'cold'.
    """
    if defaults is None:
        defaults = {
            "very_hot_Tmax_C": 30.0,
            "very_cold_Tmin_C": 5.0,
            "very_windy_speed_ms": 6.0,
            "very_wet_precip_mmday": 10.0,
            "very_uncomfortable_HI_C": 32.0,
        }

    def get(s, key):
        try:
            v = float(s.get(key, np.nan))
            return v if np.isfinite(v) else np.nan
        except Exception:
            return np.nan

    thr = {}

    if "Tmax_C" in stats:
        v = get(stats["Tmax_C"], "p90")
        thr["very_hot_Tmax_C"] = v if np.isfinite(v) else defaults["very_hot_Tmax_C"]

    if "Tmin_C" in stats:
        v = get(stats["Tmin_C"], "p10")
        thr["very_cold_Tmin_C"] = v if np.isfinite(v) else defaults["very_cold_Tmin_C"]

    if "WS_ms" in stats:
        v = get(stats["WS_ms"], "p90")
        if np.isfinite(v):
            v = max(v, wind_floor_ms)
        thr["very_windy_speed_ms"] = v if np.isfinite(v) else defaults["very_windy_speed_ms"]

    if "P_mmday" in stats:
        v = get(stats["P_mmday"], "p90")
        if np.isfinite(v):
            v = max(v, wet_floor_mm)
        thr["very_wet_precip_mmday"] = v if np.isfinite(v) else defaults["very_wet_precip_mmday"]

    if "HI_C" in stats:
        v = get(stats["HI_C"], "p90")
        thr["very_uncomfortable_HI_C"] = v if np.isfinite(v) else defaults["very_uncomfortable_HI_C"]

    return thr
```
